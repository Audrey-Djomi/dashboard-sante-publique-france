```python
import pandas as pd
import numpy as np
```

```python
# nous allons récupérer notre dataset pour néttoyer
df = pd.read_csv("../data/covid-hospit-2023-03-31-18h01.csv" , sep = ";")
print(df.head())
```

```python
# vérifier les types des données de notre BDD
df.dtypes
```

```python
# nous obervons que notr colonne "jour" est de type object, 
# nous allons convertir cette colonne en type date
df["jour"] = pd.to_datetime(df["jour"])
print(df.head())
```

```python
# vérifions la conversion de cette colonne "jour"
print(f'la colonne jour est de type: {df["jour"].dtypes}')
```

```python
# vérifions la présence des doublons
print(f" notre dataFrame contient {df.duplicated().sum()} doublons")
```

```python
# analyse des données manquantes

donnee_manquantes = df.isnull().sum().sort_values(ascending=False)
print(donnee_manquantes)
```

```python
# On fera apparaitre en pourcentage les valeurs manquantes pour avoir une meilleure représentation
per_donnee_manquantes = round(donnee_manquantes/len(df)*100) , 2
print(per_donnee_manquantes)
```
